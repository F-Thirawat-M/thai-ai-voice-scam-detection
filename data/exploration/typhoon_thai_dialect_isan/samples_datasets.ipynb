```python
from datasets import load_dataset
from IPython.display import Audio, display

# อ่านแบบ streaming ไม่โหลดทั้งชุดล่วงหน้า
ds = load_dataset(
    "typhoon-ai/thai-dialect-isan-dataset",
    revision="dfc7e495b3069940880d184f0acfcc6b3cd4f364",
    split="train",
    streaming=True,
)

samples = iter(ds)

def listen_next():
    row = next(samples)
    print("ชื่อคลิป:", row["name"])
    print("อายุ:", row["age"], "เพศ:", row["gender"])
    print("จังหวัดที่อาศัย:", row["province"])
    print("ข้อความ:", row["isan_spelling"])

    audio = row["audio"]
    display(Audio(
        audio["array"],
        rate=int(audio["sampling_rate"]),
        normalize=False,
    ))

listen_next()
```
